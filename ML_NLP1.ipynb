```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('train.txt',sep=';',header= None,names=['Text','Emotions'])
```

```python
df.isnull().sum()
```

```python
unique_emotion = df['Emotions'].unique()
emotion_number = {}
i = 0 
for emo in unique_emotion:
    emotion_number[emo]=i
    i+=1
df['Emotions']=  df['Emotions'].map(emotion_number)
```

```python
df
```

```python
df['Text'] = df['Text'].apply( lambda x : x.lower() )
```

```python
import string
```

```python
def remove_punctuation(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['Text'] = df['Text'].apply(remove_punctuation)
```

```python
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new
df['Text'] = df['Text'].apply(remove_numbers)    
```

```python
def remove_emojis(txt):
    new=''
    for i in txt:
        if i.isascii():
            new +=i
    return new

df['Text'] = df['Text'].apply(remove_emojis)        
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
stop_words
```

```python
def remove(txt):
    words = word_tokenize(txt)
    cleaned = []
    for i in words:
        if not i in stop_words:
            cleaned.append(i)
    return ' '.join(cleaned)
```

```python
df['Text'] = df['Text'].apply(remove)
```

```python
from sklearn.model_selection import train_test_split
```

```python
 X_train, X_test, y_train, y_test = train_test_split(df['Text'], df['Emotions'], test_size=0.33, random_state=42)
```

```python
X_train
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
# 1. Initialize
bow_vect = CountVectorizer()

# 2. Fit and transform the training data
X_train_bow = bow_vect.fit_transform(X_train)

# 3. ONLY transform the test data
X_test_bow = bow_vect.transform(X_test)

# 4. Now run your model
nb_model = MultinomialNB()
nb_model.fit(X_train_bow, y_train)
pred_bow = nb_model.predict(X_test_bow)
print(accuracy_score(y_test,pred_bow))
```
